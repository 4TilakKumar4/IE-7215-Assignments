```python
import matplotlib.pyplot as plt
```

```python
# Data extracted from Lab 01 Run 1 table
clock_times = [0, 0.28, 0.61, 1.23, 1.41, 2.84, 3.19, 3.65, 4.91, 5.07, 5.51, 
               6.49, 6.82, 7.61, 8.26, 8.35, 8.37, 9.14, 9.98, 10.11, 10.15]
states = [0, 1, 2, 1, 0, 1, 2, 3, 2, 3, 2, 3, 2, 1, 0, 1, 0, 1, 0, 1, 0]

# Increase the figure size (width, height in inches)
plt.figure(figsize=(12, 6))

# Creating the step plot
plt.step(clock_times, states, where='post', linewidth=2.5, color='#1f77b4')

# Formatting labels and grid
plt.xlabel('Clock Time ($t$)', fontsize=14, fontweight='bold')
plt.ylabel('Number of Patients in System ($L(t)$)', fontsize=14, fontweight='bold')
plt.title('Sample Path of Number of Patients in System ($L(t)$)', fontsize=16, pad=20)
plt.grid(True, linestyle='--', alpha=0.7)

# Setting axis limits and integer ticks for the Y-axis
plt.xticks(range(0, 12), fontsize=12)
plt.yticks(range(0, 5), fontsize=12)
plt.ylim(bottom=-0.1, top=4.5)
plt.xlim(left=0, right=11)

# Save with high DPI for a much larger, sharper file
plt.savefig('sample_path_high_res.png', dpi=800, bbox_inches='tight')
plt.show()
```
